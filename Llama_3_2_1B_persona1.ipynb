```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("text-generation", model="meta-llama/Llama-3.2-1B")
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-1B")
model = AutoModelForCausalLM.from_pretrained("meta-llama/Llama-3.2-1B", device_map="auto")
```

```python
import pandas as pd


num_rows_to_process = 4

fixed_persona = "Maria Buendia, a 34‑year‑old community‑focused event planner, leverages their bilingual fluency, methodical budgeting expertise, and passion for the arts to design inclusive festivals, workshops, and cultural programs for a local arts center, while actively pursuing certification in community development to transition into a formal outreach coordinator role."

df_input = pd.read_csv("sampled_data.csv")
if num_rows_to_process is not None:
    df_input = df_input.head(num_rows_to_process)

results_list = []

for index, row in df_input.iterrows():
    subtopic = row['subtopic']
    question = row['question']

    prompt = f"Persona: {fixed_persona}\nQuestion: {question}\nAnswer:"

    generated_answer = pipe(prompt, max_new_tokens=256, do_sample=True, return_full_text=False)[0]['generated_text'].strip()

    results_list.append({
        'subtopic': subtopic,
        'persona': fixed_persona,
        'question': question,
        'generated_output': generated_answer
    })

df_output = pd.DataFrame(results_list)

df_output.to_csv("output.csv", index=False)
```
